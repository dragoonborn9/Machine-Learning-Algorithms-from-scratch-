```python
def gradient_descent(X, iterations, alpha, y):
    theta = np.zeros((X.shape[1],1))
    h = X@theta
    for it in range(iterations):
        h = X@theta
        temp = h - y
        temp = temp*alpha
        n = theta.size
        for i in range(n):
            to_mul = X[:, i].reshape(-1,1)
            theta[i] = theta[i] - np.sum(temp*to_mul)/len(X)
    y_pred = X@theta
    error = np.sum((y_pred - y)**2)/(2*len(X)) # This is J(theta) = mean square error
    print("Error at learning rate", alpha, " = ", error)
```

```python
def LinearRegression(X, y):#using normal equation
    theta = X.T @ X
    theta = np.linalg.inv(theta)
    temp = X.T@y
    theta = theta@temp
    y_pred = X@theta
    error = np.sum((y - y_pred)**2)/(2*len(X))
    print("Error with normal equations =", error)
    
```
